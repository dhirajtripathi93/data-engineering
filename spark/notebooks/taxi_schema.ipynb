```python
df_green_sql=spark.read.parquet(f'/home/dhirajtripathi93/data-engineering/spark/data/pq/green/*/*')
```

```python
df_green_sql = df_green_sql\
                .withColumnRenamed('lpep_pickup_datetime','pickup_datetime')\
                .withColumnRenamed('lpep_dropoff_datetime','dropoff_datetime')
```

```python
df_yellow_sql = spark.read.parquet(f'/home/dhirajtripathi93/data-engineering/spark/data/pq/yellow/*/*')
```

```python
df_yellow_sql = df_yellow_sql \
    .withColumnRenamed('tpep_pickup_datetime', 'pickup_datetime') \
    .withColumnRenamed('tpep_dropoff_datetime', 'dropoff_datetime')
```

```python
common_columns =[]

yellow_columns = set(df_yellow_sql.columns)

for col in df_green_sql.columns:
    if col in yellow_columns:
        common_columns.append(col)
```

```python
from pyspark.sql import functions as F                
```

```python
df_green_sel = df_green_sql\
                    .select(common_columns)\
                    .withColumn('service_type',F.lit('green'))
```

```python
df_yellow_sel = df_yellow_sql\
                    .select(common_columns)\
                    .withColumn('service_type',F.lit('yellow'))
```

```python
df_trips_data = df_green_sel.union(df_yellow_sel)
```

```python
df_trips_data.groupBy('service_type').count().show()
```

```python
df_trips_data.registerTempTable('trips_data')
```

```python
spark.conf.set("spark.sql.repl.eagerEval.enabled", True)
df_trips_data.limit(5)
```

```python
spark.sql("""
Select service_type, count(1) from trips_data group by service_type
""").show()
```

```python
df_result = spark.sql("""
SELECT 
    -- Revenue grouping 
    PULocationID AS revenue_zone,
    date_trunc('month', pickup_datetime) AS revenue_month, 
    service_type, 

    -- Revenue calculation 
    SUM(fare_amount) AS revenue_monthly_fare,
    SUM(extra) AS revenue_monthly_extra,
    SUM(mta_tax) AS revenue_monthly_mta_tax,
    SUM(tip_amount) AS revenue_monthly_tip_amount,
    SUM(tolls_amount) AS revenue_monthly_tolls_amount,
    SUM(improvement_surcharge) AS revenue_monthly_improvement_surcharge,
    SUM(total_amount) AS revenue_monthly_total_amount,
    SUM(congestion_surcharge) AS revenue_monthly_congestion_surcharge,

    -- Additional calculations
    AVG(passenger_count) AS avg_monthly_passenger_count,
    AVG(trip_distance) AS avg_monthly_trip_distance
FROM
    trips_data
GROUP BY
    1, 2, 3
""")
```

```python
df_result.coalesce(1).write.parquet(f'/home/dhirajtripathi93/data-engineering/spark/data/report/revenue/', mode='overwrite')
```

```python
df_green_revenue = spark.sql("""
SELECT 
    date_trunc('hour', lpep_pickup_datetime) AS hour, 
    PULocationID AS zone,

    SUM(total_amount) AS amount,
    COUNT(1) AS number_records
FROM
    green
WHERE
    lpep_pickup_datetime >= '2020-01-01 00:00:00'
GROUP BY
    1, 2
""")
```

```python
df_green_revenue \
    .repartition(20) \
    .write.parquet(f'/home/dhirajtripathi93/data-engineering/spark/data/report/revenue/green', mode='overwrite')
```
